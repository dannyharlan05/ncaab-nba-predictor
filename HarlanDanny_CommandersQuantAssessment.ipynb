```python
import pandas as pd
import numpy as np
import nfl_data_py as nfl

# Optional (just makes tables look nicer in notebook)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 120)

# Load 2025 play-by-play data
df = nfl.import_pbp_data([2025])

# Quick sanity check
print("Shape:", df.shape)
df.head()
```

```python
# =========================================================
# SECOND CELL — UNDERSTAND DATA STRUCTURE
# =========================================================

print("COLUMNS:\n")
print(df.columns.tolist())

print("\nPLAY TYPE COUNTS:\n")
print(df['play_type'].value_counts(dropna=False))

print("\nDOWN VALUES:\n")
print(df['down'].value_counts(dropna=False))

print("\nPOSTEAM VALUES (first 20):\n")
print(df['posteam'].dropna().unique()[:20])
```

```python
# =========================================================
# THIRD CELL — CLEAN + FILTER DATA
# =========================================================

clean = df.copy()

# 1. Keep only run/pass plays
clean = clean[clean["play_type"].isin(["run", "pass"])]

# 2. Remove missing key values
clean = clean[
    clean["posteam"].notna() &
    clean["down"].notna() &
    clean["ydstogo"].notna() &
    clean["yards_gained"].notna() &
    clean["yardline_100"].notna()
]

# 3. Keep regular season only (important)
clean = clean[clean["season_type"] == "REG"]

# 4. Remove red zone (inside opponent 20)
clean = clean[clean["yardline_100"] > 20]

# 5. Remove backed up (own 1–10)
clean = clean[clean["yardline_100"] < 90]

# 6. Remove penalties / fake plays
clean = clean[clean["penalty"] != 1]

# 7. Remove spikes / kneels
clean = clean[clean["qb_spike"] != 1]
clean = clean[clean["qb_kneel"] != 1]

# 8. Keep only real downs
clean = clean[clean["down"].isin([1,2,3,4])]

# 9. Remove weird distances
clean = clean[clean["ydstogo"] > 0]

print("Clean shape:", clean.shape)
clean.head()
```

```python
# =========================================================
# FOURTH CELL — CREATE DOWN & DISTANCE BUCKETS
# =========================================================

def get_bucket(row):
    down = int(row["down"])
    y = row["ydstogo"]

    if down == 1:
        if y == 10:
            return "1st & 10"
        else:
            return None

    if down == 2:
        if 1 <= y <= 2:
            return "2nd & 1-2"
        elif 3 <= y <= 6:
            return "2nd & 3-6"
        elif y >= 7:
            return "2nd & 7+"

    if down == 3:
        if 1 <= y <= 2:
            return "3rd & 1-2"
        elif 3 <= y <= 5:
            return "3rd & 3-5"
        elif 6 <= y <= 9:
            return "3rd & 6-9"
        elif y >= 10:
            return "3rd & 10+"

    if down == 4:
        if 1 <= y <= 2:
            return "4th & 1-2"
        elif y >= 3:
            return "4th & 3+"

    return None

clean["bucket"] = clean.apply(get_bucket, axis=1)

# remove anything not in a bucket
clean = clean[clean["bucket"].notna()]

print(clean["bucket"].value_counts())
```

```python
def build_table(team):
    team_df = clean[clean["posteam"] == team]

    rows = []

    order = [
        "1st & 10",
        "2nd & 1-2",
        "2nd & 3-6",
        "2nd & 7+",
        "3rd & 1-2",
        "3rd & 3-5",
        "3rd & 6-9",
        "3rd & 10+",
        "4th & 1-2",
        "4th & 3+"
    ]

    for bucket in order:
        g = team_df[team_df["bucket"] == bucket]

        runs = g[g["play_type"] == "run"]
        passes = g[g["play_type"] == "pass"]

        total = len(g)

        run_ypp = f"{runs['yards_gained'].mean():.1f}" if len(runs) > 0 else "-"
        pass_ypp = f"{passes['yards_gained'].mean():.1f}" if len(passes) > 0 else "-"
        all_ypp = f"{g['yards_gained'].mean():.1f}" if total > 0 else "-"

        rows.append({
            "Down & Distance": bucket,
            "Run %": f"{round(100 * len(runs) / total)}%" if total > 0 else "0%",
            "Run Plays": len(runs),
            "Run Yds/Play": run_ypp,
            "Pass Plays": len(passes),
            "Pass Yds/Play": pass_ypp,
            "All Plays": total,
            "All Yds/Play": all_ypp
        })

    table = pd.DataFrame(rows)

    # TOTAL ROW
    total_df = team_df
    runs = total_df[total_df["play_type"] == "run"]
    passes = total_df[total_df["play_type"] == "pass"]

    total_row = pd.DataFrame([{
        "Down & Distance": "Total",
        "Run %": f"{round(100 * len(runs) / len(total_df))}%",
        "Run Plays": len(runs),
        "Run Yds/Play": f"{runs['yards_gained'].mean():.1f}",
        "Pass Plays": len(passes),
        "Pass Yds/Play": f"{passes['yards_gained'].mean():.1f}",
        "All Plays": len(total_df),
        "All Yds/Play": f"{total_df['yards_gained'].mean():.1f}"
    }])

    table = pd.concat([table, total_row], ignore_index=True)

    return table
```

```python
commanders = build_table("WAS")
eagles = build_table("PHI")
colts = build_table("IND")
```

```python
from matplotlib.backends.backend_pdf import PdfPages
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

def simple_table(df, subtitle, title_color, table_color, pdf):
    fig, ax = plt.subplots(figsize=(12, 7))
    ax.axis("off")

    # --- TITLE BAR (TEAM PRIMARY COLOR) ---
    ax.add_patch(
        Rectangle(
            (0, 0.88),
            1,
            0.12,
            transform=ax.transAxes,
            color=title_color
        )
    )

    ax.text(
        0.5, 0.94,
        "Offensive Run / Pass Summary",
        transform=ax.transAxes,
        ha="center",
        va="center",
        fontsize=16,
        color="white",
        weight="bold"
    )

    ax.text(
        0.5, 0.90,
        subtitle,
        transform=ax.transAxes,
        ha="center",
        va="center",
        fontsize=11,
        color="white",
        weight="bold"
    )

    # --- TABLE ---
    table = ax.table(
        cellText=df.values,
        colLabels=df.columns,
        cellLoc="center",
        bbox=[0, 0.05, 1, 0.83]
    )

    table.auto_set_font_size(False)
    table.set_fontsize(10)
    table.scale(1.1, 1.5)

    # style cells
    for (r, c), cell in table.get_celld().items():
        cell.set_edgecolor("black")
        cell.set_linewidth(1.0)

        # header
        if r == 0:
            cell.set_facecolor(table_color)
            cell.set_text_props(color="white", weight="bold")

        # total row
        if r == len(df):
            cell.set_facecolor(table_color)
            cell.set_text_props(color="white", weight="bold")

    # --- FOOTNOTE ---
    ax.text(
        0.5, 0.03,
        "Excludes red zone, backed up field position, penalties, and non-standard situations",
        transform=ax.transAxes,
        ha="center",
        va="center",
        fontsize=9,
        color="black"
    )

    pdf.savefig(fig, bbox_inches="tight")
    plt.close(fig)


# --- EXPORT PDF ---
with PdfPages("CommandersQuantAssessment_Part2.pdf") as pdf:
    simple_table(
        commanders,
        "Washington Commanders - 2025 Regular Season",
        title_color="#5A1414",   # burgundy
        table_color="#FFB612",   # GOLD
        pdf=pdf
)

    simple_table(
        eagles,
        "Philadelphia Eagles - 2025 Regular Season",
        title_color="#004C54",   # green
        table_color="#6C7A7D",   # silver
        pdf=pdf
    )

    simple_table(
        colts,
        "Indianapolis Colts - 2025 Regular Season",
        title_color="#002C5F",   # blue
        table_color="#5F6A6A",   # grey
        pdf=pdf
    )
```
